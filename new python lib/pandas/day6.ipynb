```python


data = {
    "Name": ["Aman", "Riya", "Rohan", "Aman", "Priya"],
    "Age": [21, 22, np.nan, 21, 24],
    "Marks": [85, 90, 78, 85, np.nan],
    "City": ["Indore", "Bhopal", "Indore", "Indore", "Delhi"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.rename(columns={"Name": "Student_Name"}, inplace=True)

print(df)
```

```python
df.rename(columns={
    "Age": "Student_Age",
    "Marks": "Student_Marks",
    "City": "Student_City"
}, inplace=True)

print(df)
```

```python
df["Student_Age"] = df["Student_Age"].astype("Int64")

print(df.dtypes)

df["Student_City"] = df["Student_City"].astype(str)
print(df.dtypes)
```

```python
df["Student_Marks"] = df["Student_Marks"].fillna(0)

print(df)
```

```python
df["Student_Marks"] = df["Student_Marks"].fillna(
    df["Student_Marks"].mean()
)
```
