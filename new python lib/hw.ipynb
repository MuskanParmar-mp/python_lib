```python


# Q2
# Create a 4x4 matrix of zeroes
# and convert all zeroes to 5 using np.where()

arr = np.zeros((4, 4), dtype=int)

arr = np.where(arr == 0, 5, arr)
print(arr)


```

```python
# Q3
# Create a 4x4 random integer matrix
# and cube all elements using broadcasting

arr = np.random.randint(1, 10, (4, 4))
arr = arr ** 3

print(arr)


```

```python
# Q4
# Create a 1D array of 10 random elements
# between 50 and 100
# > 70 -> -1
# other values -> 5

arr = np.random.randint(50, 101, 10)

arr = np.where(arr > 70, -1, 5)

print(arr)


```

```python


# Q1
arr = np.array([10, np.nan, 20, np.nan, 30,
                np.nan, 40, np.nan, 50, np.nan])


print(arr)

# A) Find mean of data
mean = np.nanmean(arr)

print("Mean:", mean)

# B) Find number of null values
null_count = np.isnan(arr).sum()

print("Number of null values:", null_count)
```

```python
# Q2

arr = np.random.randint(1, 100, (2, 4))

print(arr)


# 1. Column-wise mean
column_mean = np.mean(arr, axis=0)
print(column_mean)


# 2. Row-wise median
row_median = np.median(arr, axis=1)
print(row_median)


# 3. Standard deviation and variance
std = np.std(arr)
variance = np.var(arr)
print(std)
print(variance)


# 4. Row-wise sum
row_sum = np.sum(arr, axis=1)
print(row_sum)
```
