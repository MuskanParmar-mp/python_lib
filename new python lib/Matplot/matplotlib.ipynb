```python
import pandas as pd

data = {
    "Department": [
        "Sales", "Marketing", "HR", "IT", "Finance",
        "Operations", "Support", "Admin", "Research", "Training"
    ],
    "Employees": [45, 30, 15, 50, 25, 40, 35, 20, 18, 12],
    "Salary": [50000, 55000, 45000, 70000, 65000,
               52000, 48000, 43000, 75000, 46000]
}

# Create DataFrame
df = pd.DataFrame(data)
df
```

```python
# bar plot
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.bar(df["Department"],df["Salary"])
plt.xlabel("Department")
plt.ylabel("salary")
plt.title("department wise salary")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df["Department"],df["Salary"],color="green")
plt.xlabel("Department")
plt.ylabel("salary")
plt.title("department wise salary")

plt.xticks(rotation=45)
plt.show()
```

```python

color=["red","green","yellow","pink","orange","gray","blue","black","brown","violet"]


plt.bar(df["Department"],df["Salary"],color=color)
plt.xlabel("Department")
plt.ylabel("salary")
plt.title("department wise salary")

plt.xticks(rotation=45)

for i ,values in enumerate(df["Employees"]):
    plt.text(i,values+1,str(values))
plt.show()
```

```python
color = ["red", "green", "yellow", "pink", "orange",
         "gray", "blue", "black", "brown", "violet"]

plt.bar(df["Department"], df["Employees"], color=color)

plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Department Wise Salary")
plt.xticks(rotation=45)

for i, values in enumerate(df["Employees"]):
    plt.text(i, values + 1, str(values), ha="center")

plt.tight_layout()
plt.show()
```
