```python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [150, 200, 180, 220, 250],  # Added sample values here
    "Profit": [20, 40, 30, 60, 50]
})

# 2. Create figure level chart
plt.figure(figsize=(10, 5))
```

```python
# to create figure level chart

# plot 1
plt.subplot(1, 2, 1)
plt.pie(df["Sales"], labels=df["Month"])

# plot 2
plt.subplot(1, 2, 2)
plt.bar(df["Month"], df["Profit"])
plt.xlabel("Month")
plt.title("Month wise profit")

plt.show()
```

```python
# to create figure level chart

# plot 1
plt.subplot(1, 2, 1)
plt.pie(df["Sales"], labels=df["Month"])

# plot 2
plt.subplot(2,2,4)
plt.bar(df["Month"], df["Profit"])
plt.xlabel("Month")
plt.title("Month wise profit")
plt.show()
```

```python
# subplots
fig , ax = plt.subplots(2,2, figsize=(10,7))

ax[0,0].scatter(df["Month"],df["Sales"])

ax[0,1].bar(df["Month"],df["Profit"])

ax[0,0].set_title("rrrrrrrrrrrrrrrrrrrr")

plt.tight_layout()

plt.show()
```
