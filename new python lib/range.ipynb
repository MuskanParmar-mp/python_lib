```python
import numpy as np
```

```python
arr = np.array([12,45,67,89,80])

arr[1]
arr[-1]
arr[3]
```

```python
# 2d matrix indexing

arr2 = np.array([[12,34,55],
                [56,88,12],
                [14,90,99],
                ])

arr2
print(arr2[1,1])
print(arr2[2,0])
print(arr2[0,2])
```

```python
# fancy indexing------------when we want more then 1 element with position

arr = np.array([12,45,67,89,80])
arr[[1,3]]
arr[[0,2,4]]
arr[[0,4,2]]
```

```python
# boolean indexing
arr = np.array([12,45,67,89,80])
print(arr[arr>50])
print(arr[arr%2==0])
```
