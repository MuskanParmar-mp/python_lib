## slicing


```python
a = np.array([1,2,3,4,5])
a[2:6]
```

```python
arr = np.array([[12,14,50],
                [77,66,9],
                [10,2,4],
                ])
arr
print(arr[1:3,1])
print(arr[1:3,0:2])
print(arr[1:,1:])
print(arr[0:2, 0:])

print(arr[0: , 0: ])

```

```python
a = np.arange(0,16).reshape(4,4)
print(a)
print(a[1:2,:])
print(a[3:,:])
print(a[0:,1:3])
print(a[1:3,1:])
```

```python
arr = np.zeros((6,6),dtype=int)
arr


arr[0, :] = 1      
arr[-1, :] = 1    
arr[:, 0] = 1      
arr[:, -1] = 1     

print(arr)
```

```python
# np.where() 

ar = np.array([12,45,67,89,80])
print(ar)

print(np.where(ar>50))


print(np.where(ar>50, -1, ar))
```

```python
np.random.seed(42)                    # random value (static)

arr = np.random.randint(0, 10, (3, 3))

arr

print(np.where(arr % 2 == 0, 0, 1))
```

```python
# broadcasting 

np.random.seed(42)                            # random value (static)
arr = np.random.randint(0, 10, (3, 3))
arr

for i in arr:
    i += 10
print(arr)    


for i in np.nditer(arr):
    print(i)

```

```python
for i in np.nditer(arr):
    if i%2==0:
        print(i + 10)
```

```python
# broadcasting 

print(arr + 10)

print(arr ** 2)
```
