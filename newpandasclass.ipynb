# Data cleaning

```python
import pandas as pd
import numpy as np 
mydata = pd.DataFrame({
    'Name': ['Aasish','tanish','Prachi'],
    'Score' : [83,np.nan,95],
    'Age' : [20,-1,24],
    'Date' : ['2026-01-16']

    
})
print(mydata)
```

```python
mydata = pd.DataFrame({
    'Name': ['Aasish','tanish','Prachi'],
    'Score' : [83,np.nan,95],
    'Age' : [20,-1,24],
    'Date' : ['2026-01-16' 'not_a_date']

    
})
mean_value  = mydata['Score'].mean()
mydata .fillna({"Score" : mean_value}, inplace=True)
display(mydata)
```
